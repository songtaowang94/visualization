```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests
from io import BytesIO

# Set seed for reproducibility
np.random.seed(613)

# Generate Data Set 1 (Random)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)

# Generate Data Set 2 (One Piece Characters)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Create a figure with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

# Plot on the first axis (Scatter)
ax1.scatter(x1, y1)

# Plot on the second axis (Bar)
ax2.bar(x2, y2)

fig.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(10, 4))

# Customizing Plot 1
ax1.scatter(x1, y1, color='orange', marker='^', alpha=0.7)
ax1.set_title('Random Scatter', fontdict={'color': 'darkred'})
ax1.set_ylabel('Value')

# Customizing Plot 2
ax2.bar(x2, y2, color='teal')
ax2.set_title('Crew Power Levels', fontdict={'family': 'serif'})
ax2.grid(axis='y', linestyle='--', alpha=0.5)

fig.tight_layout()
plt.show()
```

```python
# Define layout: 'ax1' top left, 'ax3' right column (spanning both rows), 'ax2' bottom left
layout = [
    ['ax1', 'ax3'],
    ['ax2', 'ax3']
]

fig, someaxes = plt.subplot_mosaic(layout, figsize=(7, 4))

# Access axes by their string labels
someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

plt.show()
```

```python
# Create mosaic with constrained layout
fig, someaxes = plt.subplot_mosaic(layout, figsize=(7, 4), layout="constrained")

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

# Adding large labels to demonstrate how layout handles space
someaxes["ax1"].set_xlabel('A Big Label', fontsize=12)
someaxes["ax2"].set_xlabel('Another Label', fontsize=12)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=12)

plt.show()
```

```python
# Data setup
y2_line = np.array([170, 100, 90, 120, 50]) # New data for the line

fig, ax = plt.subplots(figsize=(7, 3))

# Plot Bar Chart
ax.bar(x2, y2, color="indigo", label="Bar Data")

# Plot Line Chart on top
ax.plot(x2, y2_line, color="red", label="Line Data")

plt.show()
```

```python
# 1. Calculate Standard Deviation
y2_sd = np.std(y2)

# 2. Plot
fig, ax = plt.subplots(figsize=(7, 3))

ax.plot(x2, y2, color="red")

# 3. Add Error Bars
ax.errorbar(x2, 
            y2, 
            yerr=y2_sd, 
            fmt="none",          # Don't plot markers for the data points
            ecolor="indigo",     # Error bar color
            elinewidth=4,        # Line width
            capsize=6,           # Cap size
            capthick=4           # Cap thickness
           )

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Sample Data: Daily Routine Hours (inspired by the case study in the slides)
categories = ['Sleep', 'Creative Work', 'Admin', 'Leisure', 'Exercise']
hours = [8, 6, 4, 4, 2]
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# Define patterns for each bar to ensure distinction without color
patterns = ['/', '\\', '|', '-', '+']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

bars = ax.bar(categories, hours, color=colors, edgecolor='black')

# Apply hatching (texture) to each bar
for bar, pattern in zip(bars, patterns):
    bar.set_hatch(pattern)

ax.set_title("Daily Routine (Accessible with Patterns)")
ax.set_ylabel("Hours")

plt.show()
```

```python
# Font dictionary for accessible text
# Sans-serif, dark color for contrast, and large size
accessible_font = {'family': 'sans-serif', 'color': 'black', 'size': 12}
title_font = {'family': 'sans-serif', 'color': 'black', 'size': 16, 'weight': 'bold'}

fig, ax = plt.subplots(figsize=(8, 5))

# Plotting a horizontal bar chart (often easier to read for categories)
bars = ax.barh(categories, hours, color='#444444')

# 1. Title Left Alignment
ax.set_title("Breakdown of Daily Activities", loc='left', fontdict=title_font, pad=20)

# 2. Direct Labeling (Placing values next to bars instead of a legend/axis)
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.2,       # x-position (slightly right of the bar)
            bar.get_y() + bar.get_height()/2, # y-position (centered)
            f'{width} hrs',    # The label text
            ha='left', va='center', fontdict=accessible_font)

# 3. Removing unnecessary noise (spines) to reduce cognitive load
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 4. Large tick labels
ax.tick_params(axis='both', labelsize=12)

plt.show()
```

```python
# Generating sample data for a heatmap
data = np.random.rand(10, 10)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# BAD: Jet/Rainbow (Discouraged)
ax1.imshow(data, cmap='jet')
ax1.set_title("Avoid: Rainbow/Jet", loc='left')

# GOOD: Viridis (Recommended)
# It maintains luminance contrast, making it visible even in grayscale
ax2.imshow(data, cmap='viridis')
ax2.set_title("Preferred: Viridis", loc='left')

plt.show()
```

```python
# Load the 'tips' dataset
tips = sns.load_dataset("tips")
print(tips.head())

# Set a theme (style)
sns.set_style('whitegrid') 
# Try: 'darkgrid', 'white', 'dark', 'ticks'

# Create a basic line plot
plt.figure(figsize=(10, 3))
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')

# Add titles and labels
tipgraph.set(title='Tips vs. Total Bill', 
             xlabel='Total Bill ($)', 
             ylabel='Tip Amount ($)')

plt.show()
```

```python
plt.figure(figsize=(10, 3))

# Customizing specific aesthetic elements
sns.lineplot(data=tips, 
             x='total_bill', 
             y='tip',
             color='hotpink',
             linestyle='--',
             linewidth=3,
             marker='o',
             markerfacecolor='indigo',
             markersize=8)

plt.title('Tips vs. Total Bill (Customized)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Mapping 'day' to color (hue) and 'time' to marker style
tipgraph = sns.scatterplot(data=tips, 
                           x='total_bill', 
                           y='tip', 
                           style='time', 
                           hue='day', 
                           palette=['purple', 'hotpink', 'deepskyblue', 'yellowgreen'],
                           s=100) # s controls marker size

tipgraph.set(title='Tips vs. Total Bill by Day and Time',
             xlabel='Total Bill ($)', 
             ylabel='Tip Amount ($)')

plt.show()
```

```python
# 1. Pairplot
sns.pairplot(data=tips, hue='day')
plt.show()

# 2. Relplot (FacetGrid)
# Columns separated by 'day', colors by 'sex'
sns.relplot(data=tips, 
            x="total_bill", 
            y="tip", 
            hue="sex", 
            col="day", 
            kind="scatter", 
            col_wrap=2)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Set a default style
plt.style.use('default')
```

```python
# Sample Data: Causes of Mortality (Simulated)
categories = ['Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec', 'Jan', 'Feb', 'Mar']
# Data: [Disease, Wounds, Other]
disease = [10, 15, 20, 25, 40, 50, 45, 35, 30, 20, 15, 10]
wounds = [2, 3, 5, 5, 4, 3, 2, 2, 2, 3, 4, 2]

# Angles for the circle (12 months)
theta = np.linspace(0.0, 2 * np.pi, len(categories), endpoint=False)
width = (2*np.pi) / len(categories)

# Set up the Polar Plot
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'projection': 'polar'})

# Plotting the bars
# Note: We stack them to show total mortality, similar to Nightingale's approach
p1 = ax.bar(theta, disease, width=width, color='#5da5da', alpha=0.8, label='Preventable Disease')
p2 = ax.bar(theta, wounds, width=width, bottom=disease, color='#f15854', alpha=0.8, label='Wounds')

# Formatting
ax.set_xticks(theta)
ax.set_xticklabels(categories)
ax.set_title("Diagram of the Causes of Mortality (Recreation)", y=1.1)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

plt.show()
```

```python
# Data: Occupations of African Americans (Simulated Inspiration)
occupations = ['Agriculture', 'Service', 'Manufacturing', 'Commerce', 'Professions']
values = [60, 25, 10, 3, 2]

# 1. Set the "Du Bois" Aesthetic
with plt.rc_context({
    'axes.facecolor': '#e3dcd2',       # Parchment background
    'figure.facecolor': '#e3dcd2',     # Parchment background
    'font.family': 'serif',            # Serif font
    'axes.edgecolor': 'black',         # Strong borders
    'axes.grid': False                 # No grid (usually)
}):
    
    fig, ax = plt.subplots(figsize=(6, 8))
    
    # Plot Horizontal Bar (often used by Du Bois)
    # Colors: Black, Gold (#ffd700), Crimson (#dc143c)
    ax.barh(occupations, values, color=['#000000', '#dc143c', '#ffd700', '#228b22', '#000080'])
    
    # Titles and Labels
    ax.set_title("OCCUPATIONS OF NEGROES AND WHITES IN GEORGIA.", fontsize=14, weight='bold', pad=20)
    ax.set_xlabel("PERCENTAGE", fontsize=10)
    
    # Invert y-axis to have top value at the top
    ax.invert_yaxis()
    
    plt.show()
```
